```python
# Import required libraries

import cv2
import numpy as np
import matplotlib.pyplot as plt

from google.colab import files
from google.colab.patches import cv2_imshow
```

```python

# Upload and read the original noisy image

uploaded = files.upload()

image_name = list(uploaded.keys())[0]
img = cv2.imread(image_name, cv2.IMREAD_GRAYSCALE)

print("Image Shape:", img.shape)
cv2_imshow(img)
```

```python

# Apply a 3x3 averaging filter to reduce image noise

mean_3x3 = cv2.blur(img, (3, 3))

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray", vmin=0, vmax=255)
plt.title("Original Noisy Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(mean_3x3, cmap="gray", vmin=0, vmax=255)
plt.title("3x3 Averaging Filter")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python

# Apply a 3x3 median filter

median_3x3 = cv2.medianBlur(img, 3)

# Compare the original image and both smoothing filters

images = [img, mean_3x3, median_3x3]
titles = ["Original Noisy Image",
          "3x3 Averaging Filter",
          "3x3 Median Filter"]

plt.figure(figsize=(15, 5))

for i in range(3):
    plt.subplot(1, 3, i + 1)
    plt.imshow(images[i], cmap="gray", vmin=0, vmax=255)
    plt.title(titles[i])
    plt.axis("off")

plt.tight_layout()
plt.show()
```

The averaging filter reduces image noise but introduces blurring, particularly around edges and fine details. The median filter preserves edge structures better, although some noise remains visible. The results demonstrate the trade-off between noise reduction and detail preservation.
