```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

from google.colab import files
from google.colab.patches import cv2_imshow
```

```python
uploaded = files.upload()
```

```python
image_name = list(uploaded.keys())[0]
img = cv2.imread(image_name)

print("File name:", image_name)
print("Image shape:", img.shape)

cv2_imshow(img)
```

```python
# Nearest Neighbor Interpolation
# Resize the original 320x240 image to 640x480 using the nearest neighbor interpolation method.

nearest = cv2.resize(img, (640, 480), interpolation=cv2.INTER_NEAREST)
print("Nearest Neighbor Image Shape:", nearest.shape)
cv2_imshow(nearest)
```

```python
# Bilinear Interpolation
# Resize the original 320x240 image to 640x480 using
# bilinear interpolation.
# The new pixel intensity is calculated as a weighted average
# of the four neighboring pixels.

bilinear = cv2.resize(img, (640, 480), interpolation=cv2.INTER_LINEAR)
print("Bilinear Image Shape:", bilinear.shape)
cv2_imshow(bilinear)
```

```python
# Bicubic Interpolation
# Resize the original 320x240 image to 640x480 using
# bicubic interpolation.
# Each new pixel value is estimated using 16 neighboring pixels
# and cubic polynomial interpolation.

bicubic = cv2.resize(img, (640, 480), interpolation=cv2.INTER_CUBIC)
print("Bicubic Image Shape:", bicubic.shape)
cv2_imshow(bicubic)
```

```python
# Comparison of interpolation methods
# Display the original image and the three resized images.

plt.figure(figsize=(14, 10))

# 1. Original image
plt.subplot(2, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Original Image (320x240)")
plt.axis("off")

# 2. Nearest Neighbor
plt.subplot(2, 2, 2)
plt.imshow(cv2.cvtColor(nearest, cv2.COLOR_BGR2RGB))
plt.title("Nearest Neighbor (640x480)")
plt.axis("off")

# 3. Bilinear
plt.subplot(2, 2, 3)
plt.imshow(cv2.cvtColor(bilinear, cv2.COLOR_BGR2RGB))
plt.title("Bilinear (640x480)")
plt.axis("off")

# 4. Bicubic
plt.subplot(2, 2, 4)
plt.imshow(cv2.cvtColor(bicubic, cv2.COLOR_BGR2RGB))
plt.title("Bicubic (640x480)")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Compare a magnified region of the three interpolated images

plt.figure(figsize=(15, 5))

images = [nearest, bilinear, bicubic]
titles = ["Nearest Neighbor", "Bilinear", "Bicubic"]

for i in range(3):
    region = images[i][100:200, 200:300]

    plt.subplot(1, 3, i + 1)
    plt.imshow(cv2.cvtColor(region, cv2.COLOR_BGR2RGB),
               interpolation="nearest")
    plt.title(titles[i])
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Comparison of the same region in four images

original_region = img[50:100, 100:150]
nearest_region = nearest[100:200, 200:300]
bilinear_region = bilinear[100:200, 200:300]
bicubic_region = bicubic[100:200, 200:300]

images = [original_region, nearest_region,
          bilinear_region, bicubic_region]

titles = ["Original", "Nearest Neighbor",
          "Bilinear", "Bicubic"]

plt.figure(figsize=(16, 4))

for i in range(4):
    plt.subplot(1, 4, i + 1)
    plt.imshow(cv2.cvtColor(images[i], cv2.COLOR_BGR2RGB),
               interpolation="nearest")
    plt.title(titles[i])
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
# Numerical comparison using Mean Absolute Difference (MAD)

def calculate_mad(image1, image2):
    difference = np.abs(
        image1.astype(np.float32) -
        image2.astype(np.float32)
    )
    return np.mean(difference)

print("Nearest vs Bilinear:",
      calculate_mad(nearest_region, bilinear_region))

print("Nearest vs Bicubic:",
      calculate_mad(nearest_region, bicubic_region))

print("Bilinear vs Bicubic:",
      calculate_mad(bilinear_region, bicubic_region))
```
