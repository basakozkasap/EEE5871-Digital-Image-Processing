```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

from google.colab import files
from google.colab.patches import cv2_imshow
```

```python
uploaded = files.upload()
```

```python

# Read and display the original image in grayscale

image_name = list(uploaded.keys())[0]

img = cv2.imread(image_name, cv2.IMREAD_GRAYSCALE)

print("Image shape:", img.shape)

cv2_imshow(img)
```

```python
# Plot the histogram of the original grayscale image.
# The histogram shows the number of pixels at each intensity level.

plt.figure(figsize=(15, 5))

plt.hist(img.ravel(), bins=256, range=(0, 256), color="#6B7280")

plt.xlim(50, 200)

plt.title("Original Image Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Number of Pixels")

plt.show()
```

```python

# Calculate the histogram and CDF of the original image

histogram = cv2.calcHist([img], [0], None, [256], [0, 256])

probability = histogram / histogram.sum()

cdf = np.cumsum(probability)

plt.figure(figsize=(10, 2))

plt.plot(cdf, color="#0F766E", linewidth=2)

plt.title("Cumulative Distribution Function (CDF)")
plt.xlabel("Pixel Intensity")
plt.ylabel("Cumulative Probability")

plt.xlim(0, 255)
plt.ylim(0, 1.05)
plt.grid(alpha=0.3)

plt.show()
```

```python

# Histogram Equalization using the CDF transformation

# Convert cumulative probabilities to 8-bit intensity values
lookup_table = np.round(255 * cdf).astype(np.uint8).ravel()

# Replace each original pixel with its new intensity value
equalized = lookup_table[img]

# Display original and equalized images
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(img, cmap="gray", vmin=0, vmax=255)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(equalized, cmap="gray", vmin=0, vmax=255)
plt.title("Histogram Equalized Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python

# Compare histograms before and after equalization

plt.figure(figsize=(14, 5))

# Original Image Histogram
plt.subplot(1, 2, 1)
plt.hist(img.ravel(), bins=256, range=(0, 256), color="#6B7280")
plt.title("Original Image Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Number of Pixels")
plt.xlim(0, 255)

# Equalized Image Histogram
plt.subplot(1, 2, 2)
plt.hist(equalized.ravel(), bins=256, range=(0, 256), color="#0F766E")
plt.title("Equalized Image Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Number of Pixels")
plt.xlim(0, 255)

plt.tight_layout()
plt.show()
```
